# 01. Eksploracja danych Webis-CMV-20 (ChangeMyView)
W tym notebooku sprawdzamy pobrane dane ze zbioru dyskusji Reddit r/ChangeMyView.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Wczytanie przygotowanej próbki 1000 par dyskusji
data_path = '../data/processed/cmv_persuasion_pairs_sample.jsonl'
records = []
with open(data_path, 'r', encoding='utf-8') as f:
    for line in f:
        records.append(json.loads(line))

df = pd.DataFrame(records)
print('Liczba wczytanych par dyskusji:', len(df))
df.head(3)
```

```python
# Podstawowe statystyki długości tekstów
df['op_len'] = df['op_text'].apply(lambda x: len(x.split()))
df['delta_len'] = df['delta_argument'].apply(lambda x: len(x.split()))
df['nodelta_len'] = df['nodelta_argument'].apply(lambda x: len(x.split()))

df[['op_len', 'delta_len', 'nodelta_len']].describe()
```
